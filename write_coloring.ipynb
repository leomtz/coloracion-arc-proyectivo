```python
# This script reads the coloring file for PG(2,4) and groups (a,b) pairs by color j=0,1,2

color_to_points = {0: [], 1: [], 2: []}

with open('new/affine_pg24_coloring.txt', 'r') as f:
    for line in f:
        # Each line is of the form (a:b:1): j
        parts = line.strip().split(': ')
        if len(parts) != 2:
            continue
        coords, color = parts
        color = int(color)
        # coords is like (a:b:1)
        coords = coords.strip('()')
        a, b, c = coords.split(':')
        a, b = int(a), int(b)
        if color in color_to_points:
            color_to_points[color].append((a, b))

# Print the lists for each color
for j in range(3):
    points = color_to_points[j]
    print(f"j={j}: {points}")
```
